```python
# dictionary in python
info = {
    "name": "kanu",
    "cgpa": 8.0,
    "subject":["math","science"],
    3.14:"PI"
}

print(info)
print(info[3.14])
info["cgpa"]=8.5
print(info)
```

```python
# methods of dictionary

dict_keys = list(info.keys())
print(dict_keys)

dict_vals = list(info.values())
print(dict_vals)

dict_item = list(info.items())
print(dict_item)

print(info.get("cgpa"))
print(info.get("cgpa2"))   #wrong key

info.update({
    "city" : "ahmedabad"
})
print(info)
```

```python
# set in python
s={1,2,2,3,4}
print(type(s))
print(len(s))

s.add(5)
print(s)

empty_set=set()          #creating empty set
print(type(empty_set))
```

```python
# methods of set
s1={1,2,2,3,3}
s2={3,4,5,4,5}

s.add(4)
print(s)

s.remove(2)
print(s)

s.pop()
print(s)

print(s1.intersection(s2))
print(s1.union(s2))

s.clear()
print(s)
```

```python
# practice problem
# given a list of tuples with info(name,subject):
    # 1.list all unique course
    # 2.list students enrolled in english
    # 3.create dictionary(student,set of course)
```

```python
# practice problem
info = [
    ("Alice","Math"),
    ("Bob","Science"),
    ("Alice","Science"),
    ("Charlie","Math"),
    ("BOb","Math"),
    ("Alice","english"),
    ("charlie","english")
]

unique_course=set()
enroll_in_english=set()
dict_student={}
for name,subject in info:
    unique_course.add(subject)                 #1
    
    if(subject == "english"):                  #2
        enroll_in_english.add(name)
    
    if(dict_student.get(name)==None):          #3
        dict_student.update({name:set()})
        dict_student[name].add(subject)
    else:
        dict_student[name].add(subject) 


print(unique_course)
print(enroll_in_english)
print(dict_student)
```
