```python
import json

# json format
json_str = '{"name":"kanu","isTeacher":true}'
```

```python
# json to python dictionary
py_obj=json.loads(json_str)
print(type(py_obj),py_obj)
```

```python
py_obj={
    "name":"kanu",
    "isTeacher":True
}
```

```python
# python to json
json_str=json.dumps(py_obj)
print(type(json_str),json_str)
```

```python
# load json file and make it python dictionary
with open("data.json","r")as f:
    py_obj=json.load(f)
    print(py_obj)
    print(type(py_obj))
```

```python
data={
    "name":"kanu",
     "age": 19,
     "isteacher":True
}
```

```python
# write python dictionary data to json file
with open("data.json","w")as f:
    json.dump(data,f,indent=4,sort_keys=True)
```
