```python
# practice problem to search a word in file
data=True
line=1
word="python"
with open("random2.txt","r") as d:
    while data:
        data=d.readline()
        if(word in data):
            print(f"word found at line {line}")
            break
        line+=1
```

```python
# exception handling
try :
    x =int(input("enter number"))
    data=10/x
except ZeroDivisionError :
    print(f"enter valid number ")
except ValueError:
    print(f"enter valid numbers ")
else:
    print(f"ans={data}")
finally:
    print("program is complete ")
```

```python
# list comprehensions
list1 = [-8,-5,-1,0,1,5,8]

list2 = [0 if val < 0 else val for val in list1]
print(list2)
```
