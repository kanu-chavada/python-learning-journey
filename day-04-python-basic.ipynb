```python
# sclicing
word="i study python"
print(word[2:14])
print(word[2:])
print(word[:14])
print(word[:])
print(word[:len(word)])
print(word[-7:-4])       #reverse slicing
```

```python
# formating
a=5
b=10
sum=a+b
# formating using format function
# normal formating
print("language is {}".format("python"))
print("sum is {}".format(sum))
print("sum of {} & {} is {}".format(a,b,sum))

# index based formating
print("sum of {1} & {0} is {2}".format(a,b,sum))

# value based formating
print("value of vars {d} & {e}".format(d=3,e=13))

# formating using f-string
print(f"sum of {a} & {b} is {sum}")
print(f"avg of {a} & {b} is {sum/2}")
```

```python
# methods for list
nums=[1,2,4,3]

nums.append(5)
print(nums)

nums.insert(1,1.5)   #insert(index,value)
print(nums)

nums.sort()
print(nums)

nums.sort(reverse=True)
print(nums)

nums.reverse()
print(nums)
```

```python
# loop in list
nums1=[1,2,4,3,8,7]

for val in nums1:
    print(val)

x=8
idx=0
for val in nums1:
    if (val==x):
        print(f"{x} is found at index {idx}")
        break
    idx += 1
```
