```python
f = open("sample.txt","r")   #file object   #reading[default] if we are not passing mode.

data = f.read()
print(data)
print(type(data))

f.close()
```

```python
f = open("sample.txt","r")  

data = f.readline()     #read only one line
print(data)

data = f.readline()
print(data)

print(type(data))

f.close()
```

```python
f = open("sample.txt","w")       #writing,truncates file first
f.write("TEXT TO OVERWRITE \nthe complete data.")

f.close()
```

```python
f = open("sample.txt","a")       #writing,appends at end
f.write("new text is added")

f.close()
```

```python
s = open("sample1.txt","x")        #creates new & open forwriting
s.write("new file created.")

s.close()
```

```python
f = open("sample.txt","r+")   #read,write

f.write("123")      #write at beginning of the file
print(f.read())     

f.close()
```

```python
f = open("sample.txt","a+")   #read,write

f.write("123")      #start to write at end of the file

f.close()
```

```python
f = open("sample.txt","w+")   #read,write

f.write("123")     #truncates file first,start writing beginning of file.

f.close()
```
