```python
# Conditional Statements - Example 1
age = int(input("enter age: "))

if age >= 18:           # if true/false, entire block of code is executed
    print("you can vote")
    print("you can drive")
else:
    print("you can't vote")
    print("you can't drive")
```

```python
# Example 2 - Traffic Lights
color = input("enter color: ")

if color == "red":
    print("Stop")
elif color == "yellow":
    print("Look")
elif color == "green":
    print("Go")
else:
    print("wrong color")
```

```python
# Example 3
age = int(input("enter age: "))

if age < 13:
    print("child")
elif (age >= 13 and age < 18):
    print("teenager")
else:
    print("adult")
```

```python
# Example 4 - Login System
username = input("enter username: ")
password = input("enter password: ")

if (username == "admin" and password == "pass"):
    print("login successful!")
elif username != "admin":
    print("wrong username, try again.")
else:
    print("wrong password, try again.")
```

```python
# While Loop - Example 1

# while True:        # DO NOT RUN - an infinite loop
#     print("Prime")

# Example 2 - print 1 to 5
i = 1
while i <= 5:
    print(i)
    i += 1

# Example 3 - print 5 to 1
i = 5
while i > 0:
    print(i)    
    i -= 1

# Multiplication table of N
n = int(input("enter n: "))
i = 1

while i <= 10:
    print(i * n)     
    i += 1
```

```python
# Break & Continue

# Break for multiple of 6
i = 1

while i <= 10:
    if(i % 6 == 0):
        break
    print(i)           # 1, 2, 3, 4, 5, break
    i += 1

# Skip multiples of 3
i = 0

while(i < 10):
    i += 1
    if(i % 3 == 0):
        continue;      # 1, 2, continue, 4, 5, continue, 7, 8, continue, 10
    print(i)

# Print odd nums from 1 to 10 using continue
i = 0

while(i < 10):
    i += 1
    if(i % 2 == 0):
        continue;      # 1, 3, 5, 7, 9
    print(i)
```
